```python
from datasets import Dataset

def create_dpo_dataset(prompt_answer_pairs):
    """
    Convert list of (prompt, answer) pairs into DPO format for HuggingFace datasets.
    
    Args:
        prompt_answer_pairs: List of tuples [(prompt, answer), ...] 
                            where answer is either "STAY" or "SWERVE"
    
    Returns:
        datasets.Dataset: Dataset in DPO format with prompt, chosen, rejected
    """
    dpo_data = {
        "prompt": [],
        "chosen": [],
        "rejected": []
    }
    
    for prompt, answer in prompt_answer_pairs:
        # Validate answer
        answer = answer.strip().upper()
        if answer not in ["STAY", "SWERVE"]:
            raise ValueError(f"Answer must be 'STAY' or 'SWERVE', got: {answer}")
        
        # Determine rejected answer (complement)
        rejected_answer = "SWERVE" if answer == "STAY" else "STAY"
        
        # Format as conversational messages
        dpo_data["prompt"].append([
            {"role": "user", "content": prompt}
        ])
        dpo_data["chosen"].append([
            {"role": "assistant", "content": answer}
        ])
        dpo_data["rejected"].append([
            {"role": "assistant", "content": rejected_answer}
        ])
    
    # Create HuggingFace dataset
    dataset = Dataset.from_dict(dpo_data)
    
    return dataset
```
